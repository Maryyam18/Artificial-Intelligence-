```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
########## Exercise 1

import numpy as np

x = np.linspace(0, 10, 100)  
y = x  # Linear function y = x
fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])
ax.plot(x, y, 'b')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Linear Graph: y = x')
plt.show()
```

```python
########## Exercise 2

import numpy as np

fig = plt.figure()
ax1 = fig.add_axes([0, 0, 1, 1])       
ax2 = fig.add_axes([0.2, 0.5, 0.2, 0.2])  
x = np.linspace(0, 100, 100)
y = x * 2
ax1.plot(x, y, 'brown')
ax2.plot(x, y, 'brown')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
plt.show()
```

```python
########## Exercise 3

import numpy as np


x = np.linspace(0, 25, 100)
y = x**2
fig = plt.figure()
main_ax = fig.add_axes([0, 0, 1, 1])
main_ax.plot(x, y, 'b')
main_ax.set_xlabel('X')
main_ax.set_ylabel('Y')
inset_ax = fig.add_axes([0.2, 0.5, 0.4, 0.4])
inset_ax.plot(x, y, 'b')
inset_ax.set_xlim(20, 22)  
inset_ax.set_ylim(35, 50)  
inset_ax.set_title('zoom')
plt.show()
```

```python
########## Exercise 4

import numpy as np

x = np.linspace(0, 100, 100)
y = x ** 2
z = x * 2
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))
axes[0].plot(x, z, 'b', linestyle='-', linewidth=3)
axes[0].set_xlabel('x')
axes[0].set_ylabel('z')
axes[1].plot(x, y, 'r', linestyle='--', linewidth=3)
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
plt.show()
```
